## AWR2243 Configuration

### Configured Parameters

- Mode: 2-TX × 4-RX TDM-MIMO
- TX sequence: TX0, TX2
- Start frequency: 77 GHz
- Frequency slope: 18.008 MHz/µs
- ADC samples: 128
- Sampling rate: 10 Msps
- ADC start time: 2.76 µs
- Ramp end time: 16.08 µs
- Idle time: 5 µs
- Chirp loops per frame: 128
- Physical chirps per frame: 256
- Frame period: 100 ms
- Number of frames: 300
- Recording duration: 30 s

### Achievable Performance

- Range interval: approximately 0–83 m
- Range resolution: approximately 0.65 m
- Unambiguous radial velocity: approximately ±83 km/h
- Velocity resolution: approximately 1.3 km/h
- Radar frame rate: 10 Hz
- Virtual array size: 8 channels
- Raw capture size: 314,572,800 / 2 bytes
